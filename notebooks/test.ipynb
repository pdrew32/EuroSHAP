```python
import torch
import numpy as np
import pandas as pd

from torch.utils.data import TensorDataset, DataLoader
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=1000, n_features=20, n_informative=15, n_classes=2, random_state=42)

X_tensor = torch.tensor(X, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.long)

dataset = TensorDataset(X_tensor, y_tensor)
data_loader = DataLoader(dataset=dataset, batch_size=32, shuffle=True)
```

```python
for batch_X, batch_y in data_loader:
    print('features batch shape:', batch_X.shape)
    print('labels batch shape:', batch_y.shape)
    break # just print one batch
```

```python
if torch.accelerator.is_available():
    X_tensor = X_tensor.to(torch.accelerator.current_accelerator())
    y_tensor = y_tensor.to(torch.accelerator.current_accelerator())
```
